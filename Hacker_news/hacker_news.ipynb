```python
import pandas as pd, re

# Load the dataset
hn = pd.read_csv("hacker_news.csv")

# View the first few rows (optional for inspection)
hn.head()
```

```python
python_mentions = 0
pattern = r'[Pp]ython'

for items in hn['title']:
    if re.search(pattern, items):
        python_mentions += 1
```

```python
python_mentions =hn['title'].str.contains(pattern,regex=True).sum()
```

```python
pattern = r'[Rr]uby'
titles = hn['title']
ruby_titles =titles[titles.str.contains(pattern,regex=True)]
```
